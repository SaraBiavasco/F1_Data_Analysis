```python
qualifying_position = session.results[['Abbreviation', 'Position']].copy()

qualifying_position = qualifying_position.rename(
    columns={'Abbreviation': 'Driver'}
)

qualifying_position
```

Chi è stato più veloce nei singoli settori?

```python
fastest_laps_final['Sector1Seconds'] = (
    fastest_laps_final['Sector1Time'].dt.total_seconds()
)

fastest_laps_final['Sector2Seconds'] = (
    fastest_laps_final['Sector2Time'].dt.total_seconds()
)

fastest_laps_final['Sector3Seconds'] = (
    fastest_laps_final['Sector3Time'].dt.total_seconds()
) #trasformiamo i tempi in secondi
```

```python
fastest_laps_final[
    ['Driver', 'Sector1Seconds', 'Sector2Seconds', 'Sector3Seconds']
]
```

```python
fastest_s1 = fastest_laps_final.loc[
    fastest_laps_final['Sector1Seconds'].idxmin()
]

fastest_s2 = fastest_laps_final.loc[
    fastest_laps_final['Sector2Seconds'].idxmin()
]

fastest_s3 = fastest_laps_final.loc[
    fastest_laps_final['Sector3Seconds'].idxmin()
]
```

Adesso al posto dello scatter voglio trasformare il tracciato in una linea continua colorata in funzione della velocità

```python
from matplotlib.collections import LineCollection
from matplotlib.colors import Normalize
```

```python
import numpy as np

x = telemetry['X'].to_numpy()
y = telemetry['Y'].to_numpy()
speed = telemetry['Speed'].to_numpy()

points = np.array([x, y]).T.reshape(-1, 1, 2)

segments = np.concatenate(
    [points[:-1], points[1:]],
    axis=1
)
```

Ora che ho segments, devo dire a Matplotlib come colorarli.

```python
norm = Normalize(
    telemetry['Speed'].min(),
    telemetry['Speed'].max()
)

lc = LineCollection(
    segments,
    cmap='viridis',
    norm=norm
)

lc.set_array(speed)
lc.set_linewidth(5)
```

```python
fig, ax = plt.subplots(figsize=(9, 8))

ax.add_collection(lc)

ax.set_xlim(x.min(), x.max())
ax.set_ylim(y.min(), y.max())

ax.set_aspect('equal')
ax.axis('off')

fig.colorbar(
    lc,
    ax=ax,
    label='Speed (km/h)'
)

ax.set_title('Verstappen Fastest Lap - Speed Map')

plt.show()
```
